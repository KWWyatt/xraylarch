# UV-Vis + EXAFS Spectroscopy Figure

This notebook leaves `Spectroscopy_Figure.ipynb` untouched and runs the combined UV-Vis/EXAFS publication figure workflow from a separate helper module.


```python
from pathlib import Path
from IPython.display import display

from spectroscopy_figure_workflow import generate_spectroscopy_figure

REPO_ROOT = Path.cwd()
RESULT = generate_spectroscopy_figure(repo_root=REPO_ROOT)

summary_columns = [
    "folder",
    "report_file",
    "concentration_m",
    "r_factor",
    "aic",
    "n_i",
    "n_o",
    "pb_i_r",
    "pb_o_r",
    "sigma2_i",
    "sigma2_o",
]

display(RESULT["summary"][summary_columns].round(4))
print(f"EXAFS input folder: {RESULT['input_folder']}")
print(f"UV-Vis folder:      {RESULT['uv_vis_folder']}")
print(f"Output folder:      {RESULT['output_folder']}")
print(f"Parsed spectra for stacked EXAFS panel: {RESULT['spectra_count']}")
print(f"UV-Vis fit R^2 values: 1 wt% = {RESULT['fit_1']['r2']:.4f}, 10 wt% = {RESULT['fit_10']['r2']:.4f}")
print(f"Saved figure: {RESULT['png_path']}")
print(f"Saved figure: {RESULT['pdf_path']}")
print(f"Saved summary CSV: {RESULT['summary_csv_path']}")
RESULT["figure"]
```
