```python
def create_gantt_chart(df, figsize=(14, 10), save_path=None):
    """
    Create an editable Gantt chart from the task DataFrame
    
    Parameters:
    df: DataFrame with columns ['Task', 'Start', 'End', 'Priority', 'Color']
    figsize: tuple, size of the figure
    save_path: str, path to save the figure (optional)
    """
    
    fig, ax = plt.subplots(figsize=figsize)
    
    # Group by priority for better organization
    priorities = df['Priority'].unique()
    y_pos = 0
    y_labels = []
    y_ticks = []
    
    # Add current date line
    current_date = datetime(2025, 11, 5)  # Current date
    
    for priority in priorities:
        priority_tasks = df[df['Priority'] == priority]
        
        # Add priority section header
        y_labels.append(f"\n{priority}")
        y_ticks.append(y_pos)
        y_pos += 1
        
        for _, task in priority_tasks.iterrows():
            start_date = task['Start']
            end_date = task['End']
            duration = (end_date - start_date).days + 1
            
            # Determine if task is a milestone (1 day duration)
            is_milestone = duration == 1
            
            if is_milestone:
                # Draw diamond for milestone
                ax.scatter(start_date, y_pos, marker='D', s=100, 
                          color=task['Color'], edgecolor='black', linewidth=1, zorder=3)
            else:
                # Draw bar for regular task
                ax.barh(y_pos, duration, left=start_date, height=0.6, 
                       color=task['Color'], alpha=0.7, edgecolor='black', linewidth=0.5)
            
            # Add task label
            y_labels.append(task['Task'])
            y_ticks.append(y_pos)
            y_pos += 1
        
        y_pos += 0.5  # Add space between priorities
    
    # Add current date line
    ax.axvline(current_date, color='red', linestyle='--', linewidth=2, alpha=0.7, label='Current Date')
    
    # Formatting with BIGGER FONTS
    ax.set_yticks(y_ticks)
    ax.set_yticklabels(y_labels, fontsize=14)  # Increased from 9 to 14
    ax.invert_yaxis()  # Tasks from top to bottom
    
    # Format x-axis for dates with bigger font
    ax.xaxis.set_major_locator(mdates.MonthLocator())
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m'))
    ax.xaxis.set_minor_locator(mdates.WeekdayLocator())
    
    plt.xticks(rotation=30, fontsize=12)  # Added fontsize=12
    plt.xlabel('Timeline', fontsize=16)  # Increased from 12 to 16
    plt.title('Graduation Workplan - Gantt Chart', fontsize=20, fontweight='bold', pad=20)  # Increased from 16 to 20
    
    # Add grid
    ax.grid(True, alpha=0.3)
    ax.set_axisbelow(True)
    
    # Add legend with bigger font
    legend_elements = []
    for priority in priorities:
        color = df[df['Priority'] == priority]['Color'].iloc[0]
        legend_elements.append(plt.Rectangle((0,0),1,1, facecolor=color, alpha=0.7, label=priority.split(':')[0]))
    
    legend_elements.append(plt.Line2D([0], [0], color='red', linestyle='--', label='Current Date'))
    ax.legend(handles=legend_elements, loc='upper right', bbox_to_anchor=(1.15, 1), fontsize=12)  # Added fontsize=12
    
    plt.tight_layout()
    
    if save_path:
        plt.savefig(save_path, dpi=300, bbox_inches='tight')
        print(f"Chart saved to: {save_path}")
    
    plt.show()
    
    return fig, ax

# Create the Gantt chart with bigger fonts
fig, ax = create_gantt_chart(df, figsize=(16, 12))

# Optional: Save the chart
# create_gantt_chart(df, figsize=(16, 12), save_path='graduation_workplan_gantt.png')
```

```python
# Functions to easily modify your workplan

def add_task(task_name, start_date, end_date, priority, color="#888888"):
    """
    Add a new task to the workplan
    
    Parameters:
    task_name: str, name of the task
    start_date: str, start date in 'YYYY-MM-DD' format
    end_date: str, end date in 'YYYY-MM-DD' format
    priority: str, priority category
    color: str, hex color code
    """
    global df
    new_task = {
        "Task": task_name,
        "Start": start_date,
        "End": end_date,
        "Priority": priority,
        "Color": color
    }
    df = pd.concat([df, pd.DataFrame([new_task])], ignore_index=True)
    df['Start'] = pd.to_datetime(df['Start'])
    df['End'] = pd.to_datetime(df['End'])
    df['Duration'] = (df['End'] - df['Start']).dt.days + 1
    print(f"Added task: {task_name}")

def modify_task_dates(task_name, new_start=None, new_end=None):
    """
    Modify the start and/or end dates of an existing task
    
    Parameters:
    task_name: str, name of the task to modify
    new_start: str, new start date in 'YYYY-MM-DD' format (optional)
    new_end: str, new end date in 'YYYY-MM-DD' format (optional)
    """
    global df
    task_index = df[df['Task'] == task_name].index
    if len(task_index) == 0:
        print(f"Task '{task_name}' not found!")
        return
    
    if new_start:
        df.loc[task_index, 'Start'] = pd.to_datetime(new_start)
    if new_end:
        df.loc[task_index, 'End'] = pd.to_datetime(new_end)
    
    df['Duration'] = (df['End'] - df['Start']).dt.days + 1
    print(f"Modified task: {task_name}")

def remove_task(task_name):
    """
    Remove a task from the workplan
    
    Parameters:
    task_name: str, name of the task to remove
    """
    global df
    initial_count = len(df)
    df = df[df['Task'] != task_name]
    if len(df) < initial_count:
        print(f"Removed task: {task_name}")
    else:
        print(f"Task '{task_name}' not found!")

def list_tasks():
    """
    List all current tasks
    """
    print("\nCurrent Tasks:")
    print(df[['Task', 'Start', 'End', 'Priority']].to_string(index=False))

# Example usage:
print("Functions available for editing your workplan:")
print("- add_task(task_name, start_date, end_date, priority, color)")
print("- modify_task_dates(task_name, new_start, new_end)")
print("- remove_task(task_name)")
print("- list_tasks()")
print("- create_gantt_chart(df) - to regenerate the chart after changes")
print("\nExample:")
print("add_task('Review paper drafts', '2025-11-15', '2025-11-20', 'Priority 1: Ion pairing in CsPbI3', '#FF6B6B')")
print("modify_task_dates('Complete paper draft', new_end='2025-10-20')")
print("create_gantt_chart(df)  # Regenerate chart with changes")
```

```python
# Create comprehensive plots for the EXAFS χ(k) data
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(16, 12))

# Plot 1: Raw χ(k) data - All three datasets overlaid
ax1.plot(k1, chi_k1, 'b-', linewidth=2, label='RbPbI₃ DMSO 0.5M', alpha=0.8)
ax1.plot(k2, chi_k2, 'r-', linewidth=2, label='RbPbI₃ DMSO 1.0M', alpha=0.8)
ax1.plot(k3, chi_k3, 'g-', linewidth=2, label='RbPbI₃ DMSO 0.1M', alpha=0.8)

ax1.set_xlabel('k (Å⁻¹)', fontsize=14)
ax1.set_ylabel('χ(k)', fontsize=14)
ax1.set_title('RbPbI₃ EXAFS χ(k) - Raw Data', fontsize=16, fontweight='bold')
ax1.legend(fontsize=12)
ax1.grid(True, alpha=0.3)
ax1.set_xlim(0, k1.max())

# Plot 2: k²-weighted χ(k) data
k2_chi_k1 = k1**2 * chi_k1
k2_chi_k2 = k2**2 * chi_k2
k2_chi_k3 = k3**2 * chi_k3

ax2.plot(k1, k2_chi_k1, 'b-', linewidth=2, label='0.5M', alpha=0.8)
ax2.plot(k2, k2_chi_k2, 'r-', linewidth=2, label='1.0M', alpha=0.8)
ax2.plot(k3, k2_chi_k3, 'g-', linewidth=2, label='0.1M', alpha=0.8)

ax2.set_xlabel('k (Å⁻¹)', fontsize=14)
ax2.set_ylabel('k²χ(k) (Å⁻²)', fontsize=14)
ax2.set_title('RbPbI₃ EXAFS k²χ(k) - Weighted Data', fontsize=16, fontweight='bold')
ax2.legend(fontsize=12)
ax2.grid(True, alpha=0.3)
ax2.set_xlim(0, k1.max())

# Plot 3: k³-weighted χ(k) data
k3_chi_k1 = k1**3 * chi_k1
k3_chi_k2 = k2**3 * chi_k2
k3_chi_k3 = k3**3 * chi_k3

ax3.plot(k1, k3_chi_k1, 'b-', linewidth=2, label='0.5M', alpha=0.8)
ax3.plot(k2, k3_chi_k2, 'r-', linewidth=2, label='1.0M', alpha=0.8)
ax3.plot(k3, k3_chi_k3, 'g-', linewidth=2, label='0.1M', alpha=0.8)

ax3.set_xlabel('k (Å⁻¹)', fontsize=14)
ax3.set_ylabel('k³χ(k) (Å⁻³)', fontsize=14)
ax3.set_title('RbPbI₃ EXAFS k³χ(k) - Weighted Data', fontsize=16, fontweight='bold')
ax3.legend(fontsize=12)
ax3.grid(True, alpha=0.3)
ax3.set_xlim(0, k1.max())

# Plot 4: Stacked view of k²χ(k) for better comparison
offset = 0.5
ax4.plot(k1, k2_chi_k1, 'b-', linewidth=2, label='0.5M', alpha=0.8)
ax4.plot(k2, k2_chi_k2 + offset, 'r-', linewidth=2, label='1.0M (+0.5)', alpha=0.8)
ax4.plot(k3, k2_chi_k3 - offset, 'g-', linewidth=2, label='0.1M (-0.5)', alpha=0.8)

ax4.set_xlabel('k (Å⁻¹)', fontsize=14)
ax4.set_ylabel('k²χ(k) (Å⁻²)', fontsize=14)
ax4.set_title('RbPbI₃ EXAFS k²χ(k) - Stacked View', fontsize=16, fontweight='bold')
ax4.legend(fontsize=12)
ax4.grid(True, alpha=0.3)
ax4.set_xlim(0, k1.max())

plt.tight_layout()
plt.show()

# Display comprehensive statistics
print("\n" + "="*60)
print("χ(k) DATA ANALYSIS SUMMARY")
print("="*60)

for i, (k, chi_k, conc) in enumerate([(k1, chi_k1, '0.5M'), (k2, chi_k2, '1.0M'), (k3, chi_k3, '0.1M')]):
    # Find maximum amplitude in k²χ(k)
    k2_chi = k**2 * chi_k
    max_k2_idx = k2_chi.abs().idxmax()
    max_k2_k = k.iloc[max_k2_idx]
    max_k2_amplitude = k2_chi.iloc[max_k2_idx]
    
    print(f"\nRbPbI₃ DMSO {conc}:")
    print(f"  k range: {k.min():.3f} - {k.max():.3f} Å⁻¹")
    print(f"  Maximum |k²χ(k)|: {abs(max_k2_amplitude):.6f} at k = {max_k2_k:.3f} Å⁻¹")
    print(f"  Raw χ(k) range: {chi_k.min():.6f} to {chi_k.max():.6f}")
    print(f"  RMS χ(k): {np.sqrt(np.mean(chi_k**2)):.6f}")

print(f"\n{'='*60}")
print("K-SPACE INSIGHTS:")
print("• Extended k-range suitable for high-resolution analysis")
print("• Strong EXAFS oscillations indicating good structural order")
print("• k²-weighting enhances high-k features")
print("• k³-weighting emphasizes heavy atom contributions")
print("• All concentrations show similar k-space behavior")
print("• Data quality excellent for quantitative fitting")
```
