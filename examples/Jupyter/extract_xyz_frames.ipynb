# Extract Selected Frames from an XYZ Trajectory

This notebook demonstrates how to extract specific frames from a multi-frame `.xyz` trajectory file and save them as separate `.xyz` files. It uses utilities from `larch.io.xyz_frames` for parsing and writing the frames.



## Setup

Adjust the variables in the next cell to match your trajectory file, the frame numbers you wish to extract, and the desired output directory. Frame numbering is zero-based by default; set `one_based=True` if your list starts at 1.



```python
from pathlib import Path

# --- User configuration --------------------------------------------------
source_xyz = Path("examples/structuredata/struct2xas/GaBr_multi-frame.xyz")
frame_numbers = [0, 10, 25]  # replace with the frames you wish to extract
output_dir = Path("examples/structuredata/struct2xas/extracted_frames")
one_based = False  # set to True if your frame numbering starts at 1
# -------------------------------------------------------------------------

source_xyz, output_dir
```

## Extract the Frames

The cell below reads the trajectory, extracts the requested frames, and writes them to the output directory. Set `overwrite=True` if you need to replace existing files.



```python
from larch.io.xyz_frames import extract_xyz_frames

written_paths = extract_xyz_frames(
    source=source_xyz,
    frame_numbers=frame_numbers,
    output_dir=output_dir,
    one_based=one_based,
    overwrite=False,
)

written_paths
```

## Inspect Saved Frames (Optional)

Use the following cell to preview the first few lines of the extracted files and confirm that the frames look as expected.



```python
for xyz_path in written_paths:
    print(f"\n=== {xyz_path.name} ===")
    with xyz_path.open() as handle:
        for line_num, line in zip(range(5), handle):
            print(line.rstrip())
    print("...")
```
